```python
# create a column storing the medication class for each entry
df_med["med_class"] = df_med[0].apply(lambda x : x.split("-")[1])
df_med
```

```python
# the 997th entry features two dashes!
df_med[0][996]
```

```python
# the same goes for the last entry
df_med[0][999] # df_med[0][999].split(".")[1]
```

```python
# the second-to-last entry features multiple full stops
df_med[0][998]
```

```python
# splitting by "." causes issues here
df_med[0][998].split(".")
```

```python
# to account for more than a full stop, use this
df_med[0][998].split(".", maxsplit=1)
```

```python
# create a column storing the medication name (and dosage) for each entry
df_med["med_name"] = df_med[0].apply(lambda x: x.split(".", maxsplit=1)[1].split("-")[0])
df_med
```

## 02: Practical Data Manipulation and Wrangling Pt. 2

```python
# create two additional columns
# a column named 'combined', storing both the number and the medication name...
# ... and a column named 'med_class_2', storing medication names only
df_med[["combined", "med_class_2"]] = df_med[0].str.split("-", n=1, expand=True)
df_med
```

```python
# create a second data frame by filtering 'df_med' to retain only the 'med_name' and 'med_class_2' columns
df_med_2 = df_med[["med_name", "med_class_2"]]
df_med_2
```

```python
# rename the 'med_class_2' column to 'med_class'
df_med_2.rename(columns={"med_class_2": "med_class"}, inplace=True)
df_med_2
```

```python
# remove the last, problematic entry...
df_med_2 = df_med_2.drop(999)
df_med_2
```

```python
# write the data frame to a CSV file
df_med_2.to_csv("med_name_and_class.csv", index=False)
```
