## Solution

```python
# create a column storing the medication name for each entry
df_med["med_name"] = df_med[0].apply(lambda x : x.split(". ")[1].split(") - ")[0].split(" (")[0])
df_med
```

```python
# create a column storing the medication class for each entry
df_med["med_class"] = df_med[0].apply(lambda x : x.split(". ")[1].split(") - ")[1])
df_med
```

```python
df_med = df_med[["med_name", "med_class"]]
df_med
```

```python
# the (apparently) only remaining issue is at line 774
df_med["med_name"][772]
```

```python
# since this is the (apparently) only remaining issue in the whole dataset...
# ... we can apply a one-row correction
df_med.loc[772, "med_name"] = df_med.loc[772, "med_name"].replace("(25 mg", "")
df_med["med_name"][772]
```

```python
# write the data frame to a CSV file
df_med.to_csv(output_folder_path+"med_namenodosage_class.csv", index=False)
```
